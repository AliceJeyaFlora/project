```python
from gtts import gTTS
import speech_recognition as sr
from pydub import AudioSegment
from jiwer import wer
import os

# Step 1: Convert Text to Speech
def text_to_speech(text, filename="output.mp3"):
    tts = gTTS(text)
    tts.save(filename)
    print(f"[✔] Text converted to speech and saved as '{filename}'")

# Step 2: Convert MP3 to WAV (required by speech_recognition)
def convert_mp3_to_wav(mp3_file="output.mp3", wav_file="output.wav"):
    audio = AudioSegment.from_mp3(mp3_file)
    audio.export(wav_file, format="wav")
    print(f"[✔] Audio converted to WAV format as '{wav_file}'")

# Step 3: Convert Speech to Text
def speech_to_text(filename="output.wav"):
    recognizer = sr.Recognizer()
    with sr.AudioFile(filename) as source:
        audio = recognizer.record(source)
    try:
        result = recognizer.recognize_google(audio)
        return result
    except sr.UnknownValueError:
        return "[Error] Could not understand the audio."
    except sr.RequestError:
        return "[Error] API unavailable or request failed."

# Step 4: Calculate Accuracy using WER
def evaluate_accuracy(original_text, recognized_text):
    error = wer(original_text.lower(), recognized_text.lower())
    accuracy = 1 - error
    return accuracy * 100  # percentage

# --- Run the process ---
input_text = "Artificial intelligence is changing the world."

# Generate speech from text
text_to_speech(input_text)

# Convert to WAV for recognition
convert_mp3_to_wav("output.mp3", "output.wav")

# Recognize speech from WAV file
recognized = speech_to_text("output.wav")

# Output and accuracy
print("\n🎙️ Original Text:", input_text)
print("📝 Recognized Text:", recognized)

if not recognized.startswith("[Error]"):
    acc = evaluate_accuracy(input_text, recognized)
    print(f"✅ Accuracy: {acc:.2f}%")
else:
    print("⚠️ Could not calculate accuracy due to recognition error.")
```
