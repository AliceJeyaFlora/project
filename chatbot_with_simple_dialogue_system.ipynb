```python
# Simulated ChatBot for Google Colab

def chatbot_colab(user_inputs):
    responses = []
    responses.append("ChatBot: Hi there! I'm ChatBot. Type 'exit' anytime to end the chat.")

    for user_input in user_inputs:
        user_input = user_input.lower()
        responses.append(f"You: {user_input}")

        if user_input == "exit":
            responses.append("ChatBot: Goodbye! Have a great day!")
            break
        elif "hello" in user_input or "hi" in user_input:
            responses.append("ChatBot: Hello! How can I help you today?")
        elif "how are you" in user_input:
            responses.append("ChatBot: I'm just a bot, but I'm doing great! Thanks for asking.")
        elif "your name" in user_input:
            responses.append("ChatBot: I'm a simple chatbot created with Python.")
        elif "help" in user_input:
            responses.append("ChatBot: Sure! I can help you with general questions. What do you need?")
        elif "bye" in user_input:
            responses.append("ChatBot: Bye! Feel free to chat with me anytime.")
            break
        else:
            responses.append("ChatBot: I'm not sure how to respond to that. Can you ask something else?")

    return responses

# Example usage
user_inputs = [
    "Hi",
    "What's your name?",
    "How are you?",
    "Help",
    "Bye"
]

# Run the chatbot
for response in chatbot_colab(user_inputs):
    print(response)
```

```python
import gradio as gr

def chatbot_response(message, chat_history):
    message_lower = message.lower()

    if message_lower == "exit":
        reply = "Goodbye! Have a great day!"
    elif "hello" in message_lower or "hi" in message_lower:
        reply = "Hello! How can I help you today?"
    elif "how are you" in message_lower:
        reply = "I'm just a bot, but I'm doing great! Thanks for asking."
    elif "your name" in message_lower:
        reply = "I'm a simple chatbot created with Python."
    elif "help" in message_lower:
        reply = "Sure! I can help you with general questions. What do you need?"
    elif "bye" in message_lower:
        reply = "Bye! Feel free to chat with me anytime."
    else:
        reply = "I'm not sure how to respond to that. Can you ask something else?"

    chat_history.append((message, reply))
    return "", chat_history

with gr.Blocks(title="Simple ChatBot") as demo:
    gr.Markdown("## 🤖 Simple Python ChatBot")

    chatbot = gr.Chatbot()
    msg = gr.Textbox(placeholder="Type your message here...", label="Your Message")
    clear = gr.Button("Clear Chat")

    state = gr.State([])

    msg.submit(chatbot_response, [msg, state], [msg, chatbot])
    clear.click(lambda: [], None, chatbot)

demo.launch()
```
